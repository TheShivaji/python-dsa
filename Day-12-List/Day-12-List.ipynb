## 📝 01 Maximum Element In A List
`File: 01_maximum_element_in_a_list.py`


```python
def maximum_element(list):
    num = list[0]
    for i in list:
        if i > num:
            num = i
    return num


# lst = [3, 5, 2, 9, 6]
lst = [-1, -2, -3, -4]
print(maximum_element(lst))
```

## 📝 02 Sum Of Elements In A List
`File: 02_sum_of_elements_in_a_list.py`


```python
def sum_element(lst):
    sum = 0
    for i in lst:
        sum += i

    return sum


lst = [-1, -2, -3, -4]
print(sum_element(lst))
```

## 📝 03 Palindrome List
`File: 03_palindrome_list.py`


```python
def palindrome(lst):
    start = 0
    end = len(lst) - 1

    while start < end:
        if lst[start] != lst[end]:
            return False
        start += 1
        end -= 1

    return True


lst = [7, 8, 9, 8, 7]
print(palindrome(lst))
```

## 📝 04 Reverse A List
`File: 04_reverse_a_list.py`


```python
def reverse_list(lst):
    reverse_lst = []
    for i in range(len(lst) - 1, -1, -1):
        reverse_lst.append(lst[i])

    return reverse_lst


lst = [1, 2, 3, 4, 5]
print(reverse_list(lst))
```

## 📝 05 Rotate List
`File: 05_rotate_list.py`


```python
def rotate_list(arr, d):
    d = d % len(arr)

    return arr[d:] + arr[:d]


arr = [1, 2, 3, 4, 5]
d = 2

print(rotate_list(arr, d))
```
