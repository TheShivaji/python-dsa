```python
#Checking if a Path is a File or Directory
import os

path = 'example.txt'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' is neither a file nor a directory.")
```

```python
## Getting the absolute path
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```

## 🟡 Level 2: Path Inspection & Metadata

### 📝 Question 4: Separate Files and Directories (`os.path.isfile` vs `os.path.isdir`)
**Task:** Ek function banao `classify_directory(dir_path)` jo us directory ke har item ko inspect kare aur ek dictionary return kare:
`{'files': [...], 'directories': [...]}`.
Isko `'.'` (current directory) par test karo.

```python
# Q4: Classify items into files and directories
import os

def classify_directory(dir_path):
    # TODO: Implement this function
    result = {'files':[] , 'directories':[]}


    for item in os.listdir(dir_path):

        full_path = os.path.join(dir_path , item)

        if os.path.isfile(full_path):
            result["files"].append(item)
        elif os.path.isdir(full_path):
            result['directories'].append(item)
            
    return result


# Test the function:
print(classify_directory('.'))
```

```python
# Q6: Check file size in bytes and KB using os.path.getsize()
import os

import os

def get_file_size_info(filepath):
    
    if not os.path.exists(filepath):
        return f"This file does't exist"
    
    
    size_bytes = os.path.getsize(filepath)
    
    
    size_kb = round(size_bytes / 1024, 2)
    
    
    return {
        'filepath': filepath,
        'size_bytes': size_bytes,
        'size_kb': size_kb
    }


test_file = "sample_test.txt"
with open(test_file, "w") as f:
    f.write("it's easy to learn python")


print(get_file_size_info(test_file))


print(get_file_size_info("fake_file.txt"))


if os.path.exists(test_file):
    os.remove(test_file)

```
