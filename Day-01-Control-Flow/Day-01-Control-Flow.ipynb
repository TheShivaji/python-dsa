## 📝 02 Largest Of Three
`File: 02_largest_of_three.py`


```python
# Largest of Three Numbers

a = 10
b = 15
c = 20

if a >= b and a >= c:
    print(f"{a} is the Largest Number")
elif b >= a and b >= c:
    print(f"{b} is the Largest Number")
else:
    print(f"{c} is the Largest Number")
```

```python
# Student Grade Calculator

marks = 85

if marks < 0 or marks > 100:
    print("Invalid Marks")
elif marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")
```

## 📝 05 Character Checker
`File: 05_character_checker.py`


```python
# Character Checker

ch = input("Enter a character: ")

if ch.isalpha():
    if ch.lower() in ['a', 'e', 'i', 'o', 'u']:
        print("Vowel")
    else:
        print("Consonant")

elif ch.isdigit():
    print("Digit")

else:
    print("Special Character")
```

## 📝 06 Simple Calculator
`File: 06_simple_calculator.py`


```python
# Simple Calculator

num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))
operator = input("Enter Operator (+, -, *, /, %): ")

if operator == "+":
    print(num1 + num2)

elif operator == "-":
    print(num1 - num2)

elif operator == "*":
    print(num1 * num2)

elif operator == "/":
    if num2 != 0:
        print(num1 / num2)
    else:
        print("Cannot Divide by Zero")

elif operator == "%":
    if num2 != 0:
        print(num1 % num2)
    else:
        print("Cannot Modulo by Zero")

else:
    print("Invalid Operator")
```

## 📝 07 Atm Machine
`File: 07_atm_machine.py`


```python
# ATM Machine (Mini Project)

balance = 5000

print("===== ATM MENU =====")
print("1. Check Balance")
print("2. Deposit")
print("3. Withdraw")
print("4. Exit")

choice = int(input("\nEnter your choice: "))

if choice == 1:
    print(f"\nCurrent Balance: ₹{balance}")

elif choice == 2:
    deposit = int(input("Enter deposit amount: ₹"))

    if deposit <= 0:
        print("❌ Invalid Deposit Amount")
    else:
        balance += deposit
        print("✅ Deposit Successful")
        print(f"Current Balance: ₹{balance}")

elif choice == 3:
    withdraw = int(input("Enter withdraw amount: ₹"))

    if withdraw <= 0:
        print("❌ Invalid Withdraw Amount")

    elif withdraw > balance:
        print("❌ Insufficient Balance")

    else:
        balance -= withdraw
        print("✅ Withdraw Successful")
        print(f"Current Balance: ₹{balance}")

elif choice == 4:
    print("🙏 Thank You for Using Our ATM")

else:
    print("❌ Invalid Option")
```
