Collection of hands-on Python problems for **Day-02-Loops**.

## 📝 01 Print 1 To 10
`File: 01_print_1_to_10.py`


```python
# Print Numbers from 1 to 10

for i in range(1, 11):
    print(i)
```

---
## 📝 02 Reverse Count
`File: 02_reverse_count.py`


```python
# Reverse Counting (10 to 1)

for i in range(10, 0, -1):
    print(i)
```

---
## 📝 03 Even Numbers
`File: 03_even_numbers.py`


```python
# Print Even Numbers from 1 to 20

for i in range(1, 21):
    if i % 2 == 0:
        print(f"Even Number: {i}")
```

---
## 📝 04 Odd Numbers
`File: 04_odd_numbers.py`


```python
# Print Odd Numbers from 1 to 20

for i in range(1, 21):
    if i % 2 != 0:
        print(f"Odd Number: {i}")
```

---
## 📝 05 While Loop
`File: 05_while_loop.py`


```python
# Print Numbers from 1 to 10 using While Loop

i = 1

while i <= 10:
    print(i)
    i += 1
```

---
## 📝 06 Countdown
`File: 06_countdown.py`


```python
# Countdown using While Loop

i = 5

while i >= 1:
    print(i)
    i -= 1

print("Blast Off 🚀")
```

---
## 📝 07 Sum Of N Numbers
`File: 07_sum_of_n_numbers.py`


```python
# Sum of First N Numbers

number = int(input("Enter a Number: "))

total = 0

for i in range(1, number + 1):
    total += i

print(f"Sum = {total}")
```

## 📝 08 Multiplication Table
`File: 08_multiplication_table.py`


```python
# Multiplication Table

number = int(input("Enter a Number: "))

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```
