## 📝 01 Linear Search
`File: 01_linear_search.py`


```python
def linear_search(numbers, target):

    for i in range(0, len(numbers)):
        if numbers[i] == target:
            return i

    return -1


numbers = [10, 25, 7, 42, 18, 31]
target = 31

print(linear_search(numbers, target))
```

## 📝 02 Binary Search
`File: 02_binary_search.py`


```python
def binary_search(numbers, target):

    start = 0
    end = len(numbers) - 1

    while start <= end:

        mid = (start + end) // 2

        if numbers[mid] == target:
            return mid

        elif numbers[mid] < target:
            start = mid + 1

        else:
            end = mid - 1

    return -1


numbers = [3, 7, 12, 18, 25, 31, 42, 56, 70]

target = 42

print(binary_search(numbers, target))
```

## 📝 03 Bubble Sort
`File: 03_bubble_sort.py`


```python
def bubble_sort(numbers):

    for i in range(0, len(numbers)):
        for j in range(0, len(numbers) - i - 1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

    return numbers


numbers = [64, 34, 25, 12, 22, 11, 90]

print(bubble_sort(numbers))
```

## 📝 04 Selection Sort
`File: 04_selection_sort.py`


```python
def selection_sort(numbers):

    for i in range(0, len(numbers)):
        smallest_index = i
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[smallest_index]:
                smallest_index = j
        numbers[i], numbers[smallest_index] = numbers[smallest_index], numbers[i]

    return numbers


numbers = [29, 10, 14, 37, 13]

print(selection_sort(numbers))
```

## 📝 05 Insertion Sort
`File: 05_insertion_sort.py`


```python
def insertion_sort(numbers):

    for i in range(1, len(numbers)):
        key = numbers[i]
        j = i - 1

        while j >= 0 and numbers[j] > key:
            numbers[j + 1] = numbers[j]
            j -= 1

        numbers[j + 1] = key

    return numbers


numbers = [12, 11, 13, 5, 6]

print(insertion_sort(numbers))
```

## 📝 06 Searching Practice
`File: 06_searching_practice.py`


```python
def pratice_linear(numbers, target):
    return_value = []
    for i in range(0, len(numbers)):
        if numbers[i] == target:
                return_value.append(i)

    return return_value


numbers = [5, 2, 8, 2, 9, 2, 4]
target = 2
print(pratice_linear(numbers, target))
```
