## 📝 01 Square Pattern
`File: 01_square_pattern.py`


```python
n = 3
star = []

for _ in range(1 , n + 1):
    star.append("*" * n)

print(star)
```

## 📝 02 Hollow Square
`File: 02_hollow_square.py`


```python
n = 5
for i in range(n):
    row = ""
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n-1:
            row += "*"
        else:
            row += " "
    print(row)

    
```

```python
rows = 3
columns = 5

for i in range (rows):
  row = ""
  for j in range(columns):
    row+="*"

  print(row)
```

## 📝 04 Right Angled Triangle
`File: 04_right_angled_triangle.py`


```python
n = 5

for i in  range(n):
  row=""
  for j in range(n):
    if(  j <= i):
      row+='*'
    else:
      row+=' '
  print(row)
```

## 📝 06 Pyramid Pattern
`File: 06_pyramid_pattern.py`


```python
n = 7

for i in range(1, n + 1):
    space = " " * (n - i)
    star = "*" * (2 * i - 1)
    print(space + star)
```

## 📝 09 Floyds Triangle
`File: 09_floyds_triangle.py`


```python
n = 6
number = 1

for i in range(1 , n+1):
    row = ""
    for j in range(1 , n+1):
        if(j<=i):
            row += str(number) + " "
            number+=1
        else:
            row+=" "

    print(row)
```

## 📝 10 Diamond Pattern
`File: 10_diamond_pattern.py`


```python
n = 5

for i in range(1 , n+1):
    spaces = " " * (n-i)
    stars = "*" * (2 * i -1)
    print(spaces ,stars)

for i in range(n-1 , 0 , -1):
    spaces = " " * (n-i)
    stars = "*" * (2 * i -1)
    print(spaces ,stars)
```

## 📝 12 Sandglass Pattern
`File: 12_sandglass_pattern.py`


```python
n = 5

for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces , stars)

for j in range(2 , n+1):
    spaces = " " * (n - j)
    stars = "*" * (2 * j - 1)
    print(spaces , stars)
```

## 📝 15 Number Pyramid
`File: 15_number_pyramid.py`


```python
n = 5

for i in range(1, n + 1):
    row = ""

    # Spaces
    row += " " * (n - i)

    # Numbers
    for j in range(1, 2 * i):
        row += str(j)

    print(row)
```

## 📝 Revision
`File: revision.py`


```python
rows = 4
columns = 6

for i in range(rows):
    row = ""
    for j in range(columns):
        if i == 0 or j == 0 or i == rows - 1 or j == columns - 1:
            row += "*"
        else:
            row += " "

    print(row)

# Revision Q2 — Increasing Triangle

n = 5

for i in range(n):
    row = ""
    for j in range(n):
        if j <= i:
            row += "*"
    print(row)


# Revision Q3 — Pyramid

n = 5

for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)

    print(spaces ,stars)



#Revision Q4 — Inverted Pyramid

n = 5
for i in range(n , 0 , -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)

    print(spaces , stars)


#Revision Q5 — Floyd's Triangle

n = 5

number = 1;

for i in range (1 , n+1):
    row=""
    for j in range(1 , n+1):
        if j<=i:
            row+=str(number) + " "
            number=number+1
    print(row)

```
