## 📝 01 Celsius To Fahrenheit
`File: 01_celsius_to_fahrenheit.py`


```python
def fahrenheit_convert(temp):
  return (9/5 * temp) + 32


tempe = fahrenheit_convert(25)

print(tempe)




```

## 📝 02 Area Of Rectangle
`File: 02_area_of_rectangle.py`


```python
#Area of a Rectangle

def area_of_rectangle(len , brea):
  area = len * brea

  return area

area = area_of_rectangle(5 , 3)
print(float(area))
```

## 📝 04 Lift Rounds
`File: 04_lift_rounds.py`


```python
def cal_rounds(n, capacity):
    rounds = n // capacity

    if n % capacity != 0:
        rounds += 1

    return rounds


print(cal_rounds(10, 3))
print(cal_rounds(7, 4))
print(cal_rounds(12, 3))


def calculate_vans(packages, capacity):
     vans_req = packages // capacity
     if packages % capacity != 0:
        vans_req+=1

     return vans_req

print(calculate_vans(20 , 5))

```

## 📝 05 Line Equation
`File: 05_line_equation.py`


```python
def calculate_y(slope, intercept, x):
    cal = slope *  x + intercept
    return cal


print(calculate_y(2, 3, 4))
print(calculate_y(1.5, -2, 2))
```
