## 📝 04 Check Unique Elements
`File: 04_check_unique_elements.py`


```python
def check_unique(numbers):
    seen = []

    for num in numbers:
        if num in seen:
            return False

        seen.append(num)

    return True


numbers = [1, 2, 3, 4, 5]

print(check_unique(numbers))
```

```python
list1 = [1, 3, 5]
list2 = [2, 4, 6]
merged = []

i = 0
j = 0

# Compare elements from both lists
while i < len(list1) and j < len(list2):
    if list1[i] > list2[j]:
        merged.append(list2[j])
        j += 1
    else:
        merged.append(list1[i])
        i += 1


while i < len(list1):
    merged.append(list1[i])
    i += 1


while j < len(list2):
    merged.append(list2[j])
    j += 1

print(merged)
```

## 📝 08 Max Difference Consecutive Elements
`File: 08_max_difference_consecutive_elements.py`


```python
numbers = [10, 3, 8, 20, 5]

max_difference = 0

for i in range(len(numbers) - 1):
    current_dif = numbers[i] - numbers[i + 1]

    if abs(current_dif) > max_difference:
        max_difference = current_dif


print(max_difference)
```

## 📝 11 Merge Multiple Dictionaries
`File: 11_merge_multiple_dictionaries.py`


```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
dict3 = {"e": 5}

def merge_three_dictionaries(dict1, dict2, dict3):

    merged_dict = dict1.copy()


    for key, value in dict2.items():
        merged_dict[key] = value

    for key, value in dict3.items():
        merged_dict[key] = value

    return merged_dict


print(merge_three_dictionaries(dict1 , dict2 , dict3))


```

## 📝 13 Palindromic Tuple
`File: 13_palindromic_tuple.py`


```python
def is_palindromic_tuple(tup):
    print(len(tup))
    reversed_tup = ()
    for num in range(len(tup) - 1, -1, -1):
        reversed_tup += (tup[num],)

    if reversed_tup == tup:
        return True


    return False


is_palindromic_tuple((1, 2, 3, 2, 1))


print(is_palindromic_tuple((1, 2, 3, 4, 5)))
```

## 📝 14 Merge Dictionaries With Common Keys
`File: 14_merge_dictionaries_with_common_keys.py`


```python
def merge_dicts_with_overlapping_keys(dicts):
    result = {}

    for dict in dicts:
        for key , val in dict.items():
            if key in result:
                result[key] = result[key] + val
            else:
                result[key] = val
    return result





dicts = [
    {"a": 1, "b": 2},
    {"b": 3, "c": 4},
    {"c": 5, "d": 6}
]
print(merge_dicts_with_overlapping_keys(dicts))


```

## 📝 15 Check Subset Of List
`File: 15_check_subset_of_list.py`


```python
def is_subset(lst1, lst2):
    for element in lst1:
        found = False

        for item in lst2:
            if element == item:
                found = True
                break

        if not found:
            return False

    return True


lst1 = [1, 2, 3]
lst2 = [1, 2, 3, 4, 5]

print(is_subset(lst1, lst2))
```
