## 📝 01 Basic Binary Search
`File: 01_basic_binary_search.py`


```python
def binary_search(numbers, target):

    start = 0
    end = len(numbers) - 1

    while start <= end:
        mid = (start + end) // 2

        if numbers[mid] == target:
            return mid
        elif numbers[mid] > target:
            end = mid - 1
        else:
            start = mid + 1

    return -1


numbers = [3, 7, 12, 18, 25, 31, 42]
target = 25

print(binary_search(numbers, target))
```

## 📝 02 Search Target Exists
`File: 02_search_target_exists.py`


```python
def target_exists(numbers, target):

    start = 0
    end = len(numbers) - 1

    while start <= end:
        mid = (start + end) // 2

        if numbers[mid] == target:
            return True
        elif numbers[mid] < target:
            start = mid + 1
        else:
            end = mid - 1
    return False


numbers = [2, 5, 8, 12, 16, 23, 38, 45]
target = 23

print(target_exists(numbers, target))
```

## 📝 04 Target Not Found
`File: 04_target_not_found.py`


```python
def target_not(numbers, target):

    s = 0
    e = len(numbers) - 1

    while s <= e:
        m = (s + e) // 2

        if numbers[m] == target:
            return m
        elif numbers[m] > target:
            e = m - 1
            print(e)
        else:
            s = m + 1
            print(numbers[s])
    return -1


numbers = [5, 11, 18, 24, 31, 40, 52, 67]
target = 50
print(target_not(numbers, target))
```

## 📝 05 First Occurrence
`File: 05_first_occurrence.py`


```python
def first_occurrence(numbers, target):

    start = 0
    end = len(numbers) - 1

    while start <= end:
        mid = (start + end) // 2

        if numbers[mid] == target:
            first = mid
            end = mid - 1

        elif numbers[mid] < target:
            start = mid + 1
        else:
            end = mid - 1
    return first


numbers = [1, 2, 2, 2, 3, 4, 5]
target = 2

print(first_occurrence(numbers, target))
```

## 📝 11 Search Range
`File: 11_search_range.py`


```python
def search_range(numbers, target):

    start = 0
    end = len(numbers) - 1

    while start <= end:
        mid = (start + end) // 2

        if numbers[mid] > target:
            end = mid - 1
        elif numbers[mid] == target:
            return mid
        else:
            start = mid + 1
    return start


numbers = [1, 3, 5, 6]
target = 7

print(search_range(numbers, target))
```
