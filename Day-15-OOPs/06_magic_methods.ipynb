# Day 15 - Module 06: Magic / Dunder Methods## What are Magic Methods?Magic methods (dunder = **d**ouble **under**score) are special Python methods like `__init__`, `__str__` etc.They let your custom classes integrate with Python's built-in operations.## Common Magic Methods| Method | Triggered by ||:---|:---|| `__init__` | Object creation `MyClass()` || `__str__` | `print(obj)`, `str(obj)` || `__repr__` | `repr(obj)`, interactive shell || `__len__` | `len(obj)` || `__add__` | `obj1 + obj2` || `__eq__` | `obj1 == obj2` || `__lt__` | `obj1 < obj2` || `__contains__` | `item in obj` || `__getitem__` | `obj[key]` |

## Example 1: Vector Class with Operator Overloading

```python
class Vector:    def __init__(self, x, y):        self.x = x        self.y = y    def __str__(self):        return f"Vector({self.x}, {self.y})"    def __repr__(self):        return f"Vector(x={self.x}, y={self.y})"    def __add__(self, other):        return Vector(self.x + other.x, self.y + other.y)    def __sub__(self, other):        return Vector(self.x - other.x, self.y - other.y)    def __mul__(self, scalar):        return Vector(self.x * scalar, self.y * scalar)    def __len__(self):        return int((self.x**2 + self.y**2) ** 0.5)  # magnitude    def __eq__(self, other):        return self.x == other.x and self.y == other.y    def __lt__(self, other):        return len(self) < len(other)v1 = Vector(3, 4)v2 = Vector(1, 2)print(v1)print(v1 + v2)print(v1 - v2)print(v1 * 3)print(f"len(v1) = {len(v1)}")print(f"v1 == v2: {v1 == v2}")print(f"v1 > v2: {v1 > v2}")
```

## Example 2: Custom Stack with Dunder Methods

```python
class Stack:    def __init__(self):        self._data = []    def push(self, item):        self._data.append(item)    def pop(self):        if not self._data:            raise IndexError("Stack is empty")        return self._data.pop()    def __len__(self):        return len(self._data)    def __str__(self):        return f"Stack{self._data} (top={self._data[-1] if self._data else 'empty'})"    def __contains__(self, item):        return item in self._data    def __bool__(self):        return bool(self._data)    def __iter__(self):        return iter(reversed(self._data))s = Stack()s.push(10); s.push(20); s.push(30)print(s)print(f"len: {len(s)}")print(f"20 in s: {20 in s}")print(f"bool: {bool(s)}")print("Iterate (top to bottom):", list(s))print("Popped:", s.pop())
```

## Practice Questions### Q1: Fraction ClassCreate class `Fraction` with numerator and denominator.Implement:- `__str__` → `"3/4"`- `__add__` → add two fractions (use LCM logic)- `__mul__` → multiply two fractions (simplify using GCD)- `__eq__` → compare two fractions- `__lt__` → less than comparison (for sorting)Test: add `1/2 + 1/3`, multiply `2/3 * 3/4`, compare `1/2 < 3/4`

```python
# Q1 Solution: Fraction Classfrom math import gcdclass Fraction:    def __init__(self, num, den):        if den == 0:            raise ValueError("Denominator cannot be zero")        common     = gcd(abs(num), abs(den))        self.num   = num // common        self.den   = den // common        if self.den < 0:   # keep denominator positive            self.num, self.den = -self.num, -self.den    def __str__(self):        return f"{self.num}/{self.den}" if self.den != 1 else str(self.num)    def __repr__(self):        return f"Fraction({self.num}, {self.den})"    def __add__(self, other):        num = self.num * other.den + other.num * self.den        den = self.den * other.den        return Fraction(num, den)    def __mul__(self, other):        return Fraction(self.num * other.num, self.den * other.den)    def __eq__(self, other):        return self.num == other.num and self.den == other.den    def __lt__(self, other):        return self.num * other.den < other.num * self.denf1 = Fraction(1, 2)f2 = Fraction(1, 3)f3 = Fraction(2, 3)f4 = Fraction(3, 4)print(f"{f1} + {f2} = {f1 + f2}")print(f"{f3} * {f4} = {f3 * f4}")print(f"{f1} == {f2}: {f1 == f2}")print(f"{f1} < {f4}: {f1 < f4}")fractions = [f4, f1, f2, f3]fractions.sort()  # uses __lt__print("Sorted:", [str(f) for f in fractions])
```

### Q2: Shopping Cart with Dunder MethodsCreate class `Cart` with:- `__init__` — empty list of `(item, price)` tuples- `add_item(name, price)` — add to cart- `__len__` — number of items- `__str__` — formatted receipt- `__contains__` — check if item name is in cart- `__add__` — merge two carts into a new cart- `total()` — sum of all prices

```python
# Q2 Solution: Shopping Cartclass Cart:    def __init__(self):        self._items = []  # list of (name, price)    def add_item(self, name, price):        self._items.append((name, price))    def total(self):        return sum(price for _, price in self._items)    def __len__(self):        return len(self._items)    def __str__(self):        lines = [f"  {'Item':<20} {'Price':>8}",                 f"  {'-'*20} {'-------':>8}"]        for name, price in self._items:            lines.append(f"  {name:<20} Rs.{price:>6.0f}")        lines.append(f"  {'='*29}")        lines.append(f"  {'TOTAL':<20} Rs.{self.total():>6.0f}")        return "\n".join(lines)    def __contains__(self, item_name):        return any(name == item_name for name, _ in self._items)    def __add__(self, other):        merged = Cart()        merged._items = self._items + other._items        return mergedcart1 = Cart()cart1.add_item("Laptop", 55000)cart1.add_item("Mouse",  800)cart2 = Cart()cart2.add_item("Keyboard", 2500)cart2.add_item("Monitor",  15000)print(f"Laptop in cart1: {'Laptop' in cart1}")print(f"Items: {len(cart1)}")print("Cart 1:")print(cart1)combined = cart1 + cart2print("\nCombined Cart:")print(combined)
```

## Key Takeaways- Dunder methods make your objects behave like Python built-ins- `__str__` is for human-readable output; `__repr__` is for debugging- `__add__`, `__mul__` etc. enable natural math-like syntax with custom objects- `__contains__` powers the `in` operator; `__iter__` enables `for` loops
