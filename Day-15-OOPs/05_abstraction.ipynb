# Day 15 - Module 05: Abstraction## What is Abstraction?Abstraction means **hiding implementation details** and showing only the essential features.In Python, abstraction is achieved via the `abc` module (Abstract Base Classes).## Key Concepts| Term | Meaning ||:---|:---|| `ABC` | Abstract Base Class — cannot be instantiated directly || `@abstractmethod` | Forces child classes to implement this method || Interface | What the class "promises" to provide || Concrete class | A class that implements all abstract methods |```pythonfrom abc import ABC, abstractmethod```

## Example 1: Abstract Shape

````python
from abc import ABC, abstractmethodimport mathclass Shape(ABC):    @abstractmethod    def area(self) -> float:        pass    @abstractmethod    def perimeter(self) -> float:        pass    def describe(self):        # Concrete method — reusable in all children        print(f"{self.__class__.__name__}: area={self.area():.2f}, perimeter={self.perimeter():.2f}")class Circle(Shape):    def __init__(self, radius):        self.radius = radius    def area(self):        return math.pi * self.radius ** 2    def perimeter(self):        return 2 * math.pi * self.radiusclass Rectangle(Shape):    def __init__(self, w, h):        self.w, self.h = w, h    def area(self):        return self.w * self.h    def perimeter(self):        return 2 * (self.w + self.h)# Try instantiating abstract classtry:    s = Shape()except TypeError as e:    print(f"Cannot instantiate ABC: {e}")# Works fine for concrete classesshapes = [Circle(7), Rectangle(5, 8)]for s in shapes:    s.describe()
````

## Example 2: Payment Interface

````python
from abc import ABC, abstractmethodclass PaymentGateway(ABC):    @abstractmethod    def authenticate(self) -> bool:        pass    @abstractmethod    def process_payment(self, amount: float) -> str:        pass    def pay(self, amount):        if self.authenticate():            result = self.process_payment(amount)            print(f"SUCCESS: {result}")        else:            print("FAILED: Authentication error")class RazorpayGateway(PaymentGateway):    def __init__(self, api_key):        self.api_key = api_key    def authenticate(self):        return self.api_key.startswith("rzp_")    def process_payment(self, amount):        return f"Razorpay charged Rs.{amount} (fee: Rs.{amount*0.02:.0f})"class PayPalGateway(PaymentGateway):    def __init__(self, email):        self.email = email    def authenticate(self):        return "@" in self.email    def process_payment(self, amount):        return f"PayPal transferred Rs.{amount} from {self.email}"gateways = [    RazorpayGateway("rzp_live_abc123"),    PayPalGateway("user@example.com"),    RazorpayGateway("invalid_key"),]for gw in gateways:    gw.pay(500)
````

## Practice Questions### Q1: Vehicle Abstract InterfaceCreate abstract class `Vehicle(ABC)` with abstract methods:- `start_engine()` — returns a string- `fuel_consumption()` — returns km/L (float)- `max_speed()` — returns max speed in km/hConcrete classes: `Motorcycle`, `Truck`, `ElectricCar`Each must implement all 3 methods with realistic values.Create a list of one of each and print a comparison table.

````python
# Q1 Solution: Vehicle Interfacefrom abc import ABC, abstractmethodclass Vehicle(ABC):    @abstractmethod    def start_engine(self) -> str:        pass    @abstractmethod    def fuel_consumption(self) -> float:        pass    @abstractmethod    def max_speed(self) -> int:        pass    def summary(self):        print(f"{'='*50}")        print(f"  Vehicle   : {self.__class__.__name__}")        print(f"  Engine    : {self.start_engine()}")        print(f"  Efficiency: {self.fuel_consumption()} km/L")        print(f"  Max Speed : {self.max_speed()} km/h")class Motorcycle(Vehicle):    def start_engine(self):   return "Vroom! (150cc single-cylinder)"    def fuel_consumption(self): return 45.0    def max_speed(self):      return 160class Truck(Vehicle):    def start_engine(self):   return "Rumble! (4000cc diesel V8)"    def fuel_consumption(self): return 6.0    def max_speed(self):      return 120class ElectricCar(Vehicle):    def start_engine(self):   return "Silent whirr (electric motor)"    def fuel_consumption(self): return 0.0   # electric — no fuel    def max_speed(self):      return 220vehicles = [Motorcycle(), Truck(), ElectricCar()]for v in vehicles:    v.summary()
````

### Q2: Database ConnectorCreate abstract class `DatabaseConnector(ABC)` with abstract methods:- `connect()` — returns connection message- `execute(query)` — returns query result message- `disconnect()` — returns disconnection messageConcrete classes: `MySQLConnector`, `MongoDBConnector`Add a concrete method `run_query(query)` in the base class that calls all 3 in sequence.

````python
# Q2 Solution: Database Connectorfrom abc import ABC, abstractmethodclass DatabaseConnector(ABC):    @abstractmethod    def connect(self) -> str:        pass    @abstractmethod    def execute(self, query: str) -> str:        pass    @abstractmethod    def disconnect(self) -> str:        pass    def run_query(self, query):        print(self.connect())        print(self.execute(query))        print(self.disconnect())        print()class MySQLConnector(DatabaseConnector):    def __init__(self, host, db):        self.host = host        self.db   = db    def connect(self):        return f"[MySQL] Connected to {self.db} @ {self.host}"    def execute(self, query):        return f"[MySQL] Executed: '{query}' -> OK (3 rows affected)"    def disconnect(self):        return f"[MySQL] Disconnected from {self.db}"class MongoDBConnector(DatabaseConnector):    def __init__(self, cluster):        self.cluster = cluster    def connect(self):        return f"[MongoDB] Connected to cluster: {self.cluster}"    def execute(self, query):        return f"[MongoDB] Ran aggregation pipeline: '{query}' -> 7 documents"    def disconnect(self):        return f"[MongoDB] Session closed for {self.cluster}"connectors = [    MySQLConnector("localhost:3306", "users_db"),    MongoDBConnector("atlas-cluster0")]for conn in connectors:    conn.run_query("SELECT * FROM orders")
````

## Key Takeaways- `ABC` prevents direct instantiation of incomplete classes- `@abstractmethod` enforces a **contract** that all children must fulfill- Concrete methods in the ABC provide shared behavior for free- Abstraction is about **what** an object does, not **how** it does it
