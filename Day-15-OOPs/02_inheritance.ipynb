# Day 15 - Module 02: Inheritance## What is Inheritance?Inheritance allows a **child class** to reuse attributes and methods from a **parent class**.Think of it as a child inheriting traits from a parent — but they can also have their own unique traits.## Key Concepts| Term | Meaning ||:---|:---|| Parent / Base class | The class being inherited from || Child / Derived class | The class that inherits || `super()` | Calls parent class methods || Method Overriding | Child redefines a parent method || Multi-level | Chain: A → B → C || Multiple | Child inherits from 2+ parents |

## Example 1: Single Inheritance — Vehicle Hierarchy

```python
class Vehicle:    def __init__(self, brand, speed):        self.brand = brand        self.speed = speed    def describe(self):        return f"{self.brand} travels at {self.speed} km/h"    def fuel_type(self):        return "Unknown"class Car(Vehicle):    def __init__(self, brand, speed, doors):        super().__init__(brand, speed)  # reuse parent __init__        self.doors = doors    def fuel_type(self):             # override parent method        return "Petrol/Diesel"    def describe(self):        base = super().describe()    # call parent describe        return f"{base} | {self.doors} doors | Fuel: {self.fuel_type()}"class ElectricCar(Car):    def fuel_type(self):        return "Electric"v = Vehicle("Generic", 60)c = Car("Honda", 120, 4)e = ElectricCar("Tesla", 250, 4)print(v.describe())print(c.describe())print(e.describe())
```

## Example 2: Multiple Inheritance — Mixin Pattern

```python
class Flyable:    def fly(self):        return f"{self.__class__.__name__} can fly!"class Swimmable:    def swim(self):        return f"{self.__class__.__name__} can swim!"class Duck(Flyable, Swimmable):    def quack(self):        return "Quack quack!"d = Duck()print(d.fly())print(d.swim())print(d.quack())print(f"MRO: {[cls.__name__ for cls in Duck.__mro__]}")
```

## Practice Questions### Q1: Employee HierarchyCreate:- Base class `Employee` with `name`, `base_salary`, method `annual_salary()` (base * 12), method `info()`- Child class `Manager(Employee)` with `team_size`, `bonus = 50000`; override `annual_salary()` to include bonus- Child class `Developer(Employee)` with `tech_stack` (list); override `info()` to also print tech stackCreate one of each and print their info.

```python
# Q1 Solution: Employee Hierarchyclass Employee:    def __init__(self, name, base_salary):        self.name        = name        self.base_salary = base_salary    def annual_salary(self):        return self.base_salary * 12    def info(self):        print(f"Name: {self.name} | Monthly: {self.base_salary} | Annual: {self.annual_salary()}")class Manager(Employee):    BONUS = 50_000    def __init__(self, name, base_salary, team_size):        super().__init__(name, base_salary)        self.team_size = team_size    def annual_salary(self):        return super().annual_salary() + self.BONUS    def info(self):        super().info()        print(f"  Role: Manager | Team Size: {self.team_size} | Bonus: {self.BONUS}")class Developer(Employee):    def __init__(self, name, base_salary, tech_stack):        super().__init__(name, base_salary)        self.tech_stack = tech_stack    def info(self):        super().info()        print(f"  Role: Developer | Stack: {', '.join(self.tech_stack)}")emp = Employee("Ravi",    40_000)mgr = Manager("Priya",   80_000, 10)dev = Developer("Arjun", 70_000, ["Python", "FastAPI", "PostgreSQL"])emp.info(); print()mgr.info(); print()dev.info()
```

### Q2: Shape Area CalculatorCreate:- Base class `Shape` with method `area()` returning 0 and method `describe()`- Child class `Rectangle(Shape)` with `width`, `height`; override `area()`- Child class `Circle(Shape)` with `radius`; override `area()` (pi * r^2)- Child class `Triangle(Shape)` with `base`, `height`; override `area()` (0.5 * b * h)Print area of all three shapes.

```python
# Q2 Solution: Shape Area Calculatorimport mathclass Shape:    def area(self):        return 0    def describe(self):        print(f"{self.__class__.__name__}: area = {self.area():.2f}")class Rectangle(Shape):    def __init__(self, width, height):        self.width  = width        self.height = height    def area(self):        return self.width * self.heightclass Circle(Shape):    def __init__(self, radius):        self.radius = radius    def area(self):        return math.pi * self.radius ** 2class Triangle(Shape):    def __init__(self, base, height):        self.base   = base        self.height = height    def area(self):        return 0.5 * self.base * self.heightshapes = [Rectangle(5, 8), Circle(7), Triangle(6, 4)]for shape in shapes:    shape.describe()
```

## Key Takeaways- Use `super()` to call parent class `__init__` and methods- **Override** a parent method in child by redefining it with the same name- **Multiple inheritance** lets a class inherit from more than one parent- Python uses **MRO (Method Resolution Order)** to decide which method runs first
