# Day 15 - Module 04: Encapsulation## What is Encapsulation?Encapsulation means **bundling data (attributes) with the methods that operate on it**, and **restricting direct access** to some components. This protects the internal state of an object.## Access Levels in Python| Convention | Syntax | Meaning ||:---|:---|:---|| Public | `self.name` | Accessible from anywhere || Protected | `self._name` | Convention: "internal use" (not enforced) || Private | `self.__name` | Name-mangled: `_ClassName__name` (harder to access) |## PropertiesUse `@property` to create **getter** and `@name.setter` to create **setter** — controlled access.

## Example 1: Private Attributes + Property

```python
class BankAccount:    def __init__(self, owner, pin, balance=0):        self.owner    = owner        self.__pin    = pin       # private        self._balance = balance   # protected    @property    def balance(self):        return self._balance    @balance.setter    def balance(self, value):        raise AttributeError("Direct balance modification not allowed. Use deposit/withdraw.")    def deposit(self, amount):        if amount <= 0:            raise ValueError("Deposit amount must be positive")        self._balance += amount        return self._balance    def withdraw(self, amount, pin):        if pin != self.__pin:            raise PermissionError("Wrong PIN!")        if amount > self._balance:            raise ValueError("Insufficient funds")        self._balance -= amount        return self._balance    def __str__(self):        return f"Account[{self.owner}] Balance: Rs.{self.balance}"acc = BankAccount("Shivaji", 1234, 5000)acc.deposit(2000)print(acc)print(acc.withdraw(1500, 1234))# Try wrong PINtry:    acc.withdraw(100, 9999)except PermissionError as e:    print(f"Error: {e}")
```

## Example 2: Validation via Setter

```python
class Temperature:    def __init__(self, celsius):        self.celsius = celsius  # calls the setter    @property    def celsius(self):        return self.__celsius    @celsius.setter    def celsius(self, value):        if value < -273.15:            raise ValueError(f"Temperature below absolute zero: {value}")        self.__celsius = value    @property    def fahrenheit(self):        return self.__celsius * 9/5 + 32    @property    def kelvin(self):        return self.__celsius + 273.15t = Temperature(100)print(f"{t.celsius}°C = {t.fahrenheit}°F = {t.kelvin}K")try:    t.celsius = -300  # Should raise errorexcept ValueError as e:    print(f"Caught: {e}")
```

## Practice Questions### Q1: Secure Employee RecordCreate class `Employee` with:- Private `__employee_id` (set only once in `__init__`, no setter)- Protected `_salary` with a `@property` getter and setter that validates salary must be > 0- Public `name`, `department`- Method `give_raise(percent)` — increases salary by given percentage- `__str__` to show all info (mask employee_id as `EMP****`)

```python
# Q1 Solution: Secure Employee Recordclass Employee:    def __init__(self, employee_id, name, department, salary):        self.__employee_id = employee_id   # set once, no setter        self.name          = name        self.department    = department        self._salary       = 0        self.salary        = salary        # use setter for validation    @property    def employee_id(self):        return f"EMP{'*' * (len(str(self.__employee_id)) - 2)}{str(self.__employee_id)[-2:]}"    @property    def salary(self):        return self._salary    @salary.setter    def salary(self, value):        if value <= 0:            raise ValueError("Salary must be a positive number")        self._salary = value    def give_raise(self, percent):        increase     = self._salary * percent / 100        self._salary += increase        print(f"Raise of {percent}% applied. New salary: Rs.{self._salary:,.0f}")    def __str__(self):        return (f"ID: {self.employee_id} | Name: {self.name} | "                f"Dept: {self.department} | Salary: Rs.{self.salary:,.0f}")emp = Employee(10042, "Priya Sharma", "Engineering", 75_000)print(emp)emp.give_raise(15)print(emp)try:    emp.salary = -5000except ValueError as e:    print(f"Error: {e}")
```

### Q2: Product InventoryCreate class `Product` with:- Private `__stock` (cannot go below 0)- `@property` for `stock` (getter), and `restock(quantity)` method to add stock- `sell(quantity)` method — reduces stock, raises error if insufficient- `@property` for `price` with validation (price > 0)- `__str__` to display product info

```python
# Q2 Solution: Product Inventoryclass Product:    def __init__(self, name, price, stock):        self.name   = name        self.__stock = 0        self._price  = 0        self.price   = price    # validate via setter        self.restock(stock)    @property    def stock(self):        return self.__stock    @property    def price(self):        return self._price    @price.setter    def price(self, value):        if value <= 0:            raise ValueError("Price must be positive")        self._price = value    def restock(self, qty):        if qty < 0:            raise ValueError("Restock quantity cannot be negative")        self.__stock += qty        print(f"Restocked {qty} units. Total stock: {self.__stock}")    def sell(self, qty):        if qty > self.__stock:            raise ValueError(f"Only {self.__stock} units available, cannot sell {qty}")        self.__stock -= qty        print(f"Sold {qty} units. Remaining stock: {self.__stock}")    def __str__(self):        return f"Product: {self.name} | Price: Rs.{self.price} | Stock: {self.stock}"p = Product("Laptop", 55_000, 10)print(p)p.sell(3)p.restock(5)print(p)try:    p.sell(20)except ValueError as e:    print(f"Error: {e}")
```

## Key Takeaways- `__private` attributes use name mangling — accessed as `_ClassName__attr`- `@property` creates a controlled getter; `@x.setter` creates a controlled setter- Validation in setters prevents corrupt state- Encapsulation is about **protecting data integrity**, not secrecy
