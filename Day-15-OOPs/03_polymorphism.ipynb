# Day 15 - Module 03: Polymorphism## What is Polymorphism?Polymorphism means **"many forms"**. The same method name behaves differently based on the object calling it.## Types of Polymorphism in Python| Type | Description ||:---|:---|| Method Overriding | Child class redefines parent method || Duck Typing | "If it walks like a duck, it quacks like a duck" — any object with the right method works || Operator Overloading | `+`, `*`, `<` etc. behave differently for different types || Built-in Polymorphism | `len()`, `+` work on str, list, int differently |

## Practice Questions### Q1: Payment Gateway SystemCreate:- Base class `Payment` with method `pay(amount)` that prints "Processing payment..."- Child `CreditCard(Payment)` — override `pay()` to print "Paid Rs.X via Credit Card (2% fee applied)"- Child `UPI(Payment)` — override `pay()` to print "Paid Rs.X via UPI (no fee)"- Child `Wallet(Payment)` — override `pay()` to print "Paid Rs.X via Wallet (cashback 1%)"Process payment of Rs.1000 through each method using a loop.

```python
# Q1 Solution: Payment Gatewayclass Payment:    def pay(self, amount):        print("Processing payment...")class CreditCard(Payment):    def pay(self, amount):        fee   = amount * 0.02        total = amount + fee        print(f"Paid Rs.{total:.0f} via Credit Card (2% fee = Rs.{fee:.0f})")class UPI(Payment):    def pay(self, amount):        print(f"Paid Rs.{amount} via UPI (no fee)")class Wallet(Payment):    def pay(self, amount):        cashback = amount * 0.01        print(f"Paid Rs.{amount} via Wallet (cashback Rs.{cashback:.0f})")gateways = [CreditCard(), UPI(), Wallet()]for gateway in gateways:    gateway.pay(1000)
```

### Q2: Notification System (Duck Typing)Create 3 independent classes (NO inheritance needed):- `EmailNotifier` with method `send(message)` — prints "Email sent: <message>"- `SMSNotifier` with method `send(message)` — prints "SMS sent: <message>"  - `PushNotifier` with method `send(message)` — prints "Push notification: <message>"Write a function `broadcast(notifiers, message)` that sends the message through all notifiers.Test with message: "Your OTP is 4821".

```python
# Q2 Solution: Notification System — Duck Typingclass EmailNotifier:    def send(self, message):        print(f"[EMAIL] {message}")class SMSNotifier:    def send(self, message):        print(f"[SMS]   {message}")class PushNotifier:    def send(self, message):        print(f"[PUSH]  {message}")def broadcast(notifiers, message):    print(f"Broadcasting: '{message}'")    print("-" * 40)    for notifier in notifiers:        notifier.send(message)notifiers = [EmailNotifier(), SMSNotifier(), PushNotifier()]broadcast(notifiers, "Your OTP is 4821")
```

## Key Takeaways- Polymorphism = same interface, different behavior- Python's duck typing lets any object participate as long as it has the right method- A `for` loop over objects calling the same method is the classic polymorphism pattern- No need for explicit type checking — just call the method and let the object handle it
