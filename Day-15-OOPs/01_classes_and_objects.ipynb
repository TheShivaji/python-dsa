# Day 15 - Module 01: Classes and Objects## What is OOP?Object-Oriented Programming (OOP) is a paradigm where we model real-world entities as **objects**.An object is an instance of a **class** — a blueprint that defines:- **Attributes** (data/state)- **Methods** (behavior/actions)## Key Concepts| Term | Meaning ||:---|:---|| `class` | Blueprint / Template || `object` | Instance of a class || `__init__` | Constructor — runs when object is created || `self` | Reference to the current instance || Instance variable | Data unique to each object || Instance method | Function defined inside a class |

## Example 1: Simple Class — Car

```python
class Car:    def __init__(self, brand, model, year):        self.brand = brand        self.model = model        self.year  = year    def describe(self):        return f"{self.year} {self.brand} {self.model}"    def age(self):        return 2024 - self.year# Creating objects (instances)car1 = Car("Toyota", "Corolla", 2018)car2 = Car("Tesla",  "Model 3", 2022)print(car1.describe())   # 2018 Toyota Corollaprint(car2.describe())   # 2022 Tesla Model 3print(f"car1 is {car1.age()} years old")
```

## Example 2: Bank Account — Real-World Model

```python
class BankAccount:    def __init__(self, owner, balance=0):        self.owner   = owner        self.balance = balance    def deposit(self, amount):        self.balance += amount        print(f"Deposited {amount}. New balance: {self.balance}")    def withdraw(self, amount):        if amount > self.balance:            print("Insufficient funds!")        else:            self.balance -= amount            print(f"Withdrew {amount}. Remaining: {self.balance}")    def get_balance(self):        return self.balance    def __str__(self):        return f"Account[{self.owner}] = Rs.{self.balance}"# Usageaccount = BankAccount("Shivaji", 1000)account.deposit(500)account.withdraw(200)print(account)
```

## Practice Questions### Q1: Student Report CardCreate a class `Student` with:- Attributes: `name`, `roll_no`, `marks` (list of 5 subject marks)- Method `percentage()` — returns average marks- Method `grade()` — returns grade:  - `>= 90` → `'A'`, `>= 75` → `'B'`, `>= 60` → `'C'`, else `'D'`- Method `report()` — prints a formatted report cardCreate 2 student objects and print both report cards.

```python
# Q1 Solution: Student Report Cardclass Student:    def __init__(self, name, roll_no, marks):        self.name    = name        self.roll_no = roll_no        self.marks   = marks  # list of 5 subject marks    def percentage(self):        return sum(self.marks) / len(self.marks)    def grade(self):        pct = self.percentage()        if pct >= 90:   return 'A'        elif pct >= 75: return 'B'        elif pct >= 60: return 'C'        else:           return 'D'    def report(self):        print(f"{'='*35}")        print(f"  Name    : {self.name}")        print(f"  Roll No : {self.roll_no}")        print(f"  Marks   : {self.marks}")        print(f"  Average : {self.percentage():.2f}%")        print(f"  Grade   : {self.grade()}")        print(f"{'='*35}")s1 = Student("Shivaji", 101, [88, 92, 76, 85, 90])s2 = Student("Riya",    102, [55, 60, 48, 72, 58])s1.report()s2.report()
```

### Q2: Library Book TrackerCreate a class `Book` with:- Attributes: `title`, `author`, `total_copies`, `available_copies`- Method `checkout()` — issues a book if available, else prints "Not available"- Method `return_book()` — returns a book (increment available)- Method `status()` — prints availability info

```python
# Q2 Solution: Library Book Trackerclass Book:    def __init__(self, title, author, total_copies):        self.title            = title        self.author           = author        self.total_copies     = total_copies        self.available_copies = total_copies    def checkout(self):        if self.available_copies > 0:            self.available_copies -= 1            print(f"'{self.title}' issued. Remaining: {self.available_copies}")        else:            print(f"'{self.title}' is NOT available right now.")    def return_book(self):        if self.available_copies < self.total_copies:            self.available_copies += 1            print(f"'{self.title}' returned. Available: {self.available_copies}")        else:            print("All copies already in library.")    def status(self):        print(f"Book: {self.title} | Author: {self.author} | "              f"Available: {self.available_copies}/{self.total_copies}")b = Book("Python Crash Course", "Eric Matthes", 3)b.status()b.checkout()b.checkout()b.checkout()b.checkout()  # Should say not availableb.return_book()b.status()
```

## Key Takeaways- A **class** is a blueprint; an **object** is a live instance- `__init__` initializes attributes when an object is created- `self` always refers to the current instance- Methods are functions that operate on the object's own data
