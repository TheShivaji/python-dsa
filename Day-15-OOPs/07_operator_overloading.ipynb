# Day 15 - Module 07: Operator Overloading## What is Operator Overloading?Operator overloading lets you define how standard operators (`+`, `-`, `*`, `<`, `==`, etc.)behave when applied to your custom objects.## Operator → Dunder Method Map| Operator | Method | Operator | Method ||:---:|:---|:---:|:---|| `+` | `__add__` | `==` | `__eq__` || `-` | `__sub__` | `!=` | `__ne__` || `*` | `__mul__` | `<` | `__lt__` || `/` | `__truediv__` | `>` | `__gt__` || `//` | `__floordiv__` | `<=` | `__le__` || `%` | `__mod__` | `>=` | `__ge__` || `**` | `__pow__` | `[]` | `__getitem__` |

## Example 1: Matrix Class

```python
class Matrix:    def __init__(self, data):        self.data = data        self.rows = len(data)        self.cols = len(data[0])    def __str__(self):        return "\n".join(["  " + str(row) for row in self.data])    def __add__(self, other):        if self.rows != other.rows or self.cols != other.cols:            raise ValueError("Matrix dimensions must match for addition")        result = [            [self.data[i][j] + other.data[i][j]             for j in range(self.cols)]            for i in range(self.rows)        ]        return Matrix(result)    def __mul__(self, scalar):        result = [[cell * scalar for cell in row] for row in self.data]        return Matrix(result)    def __eq__(self, other):        return self.data == other.data    def __getitem__(self, index):        return self.data[index]m1 = Matrix([[1, 2], [3, 4]])m2 = Matrix([[5, 6], [7, 8]])print("M1:\n", m1)print("M2:\n", m2)print("M1 + M2:\n", m1 + m2)print("M1 * 3:\n", m1 * 3)print("m1[0]:", m1[0])print("m1 == m2:", m1 == m2)
```

## Example 2: Money Class — Currency-Safe Arithmetic

```python
class Money:    def __init__(self, amount, currency="INR"):        self.amount   = round(amount, 2)        self.currency = currency    def __str__(self):        return f"{self.currency} {self.amount:,.2f}"    def __add__(self, other):        if self.currency != other.currency:            raise ValueError(f"Cannot add {self.currency} and {other.currency}")        return Money(self.amount + other.amount, self.currency)    def __sub__(self, other):        if self.currency != other.currency:            raise ValueError("Currency mismatch")        return Money(self.amount - other.amount, self.currency)    def __mul__(self, factor):        return Money(self.amount * factor, self.currency)    def __lt__(self, other):        return self.amount < other.amount    def __eq__(self, other):        return self.amount == other.amount and self.currency == other.currencym1 = Money(1500)m2 = Money(750)print(m1 + m2)print(m1 - m2)print(m1 * 1.18)   # with 18% GSTprint(m1 > m2)prices = [Money(500), Money(1200), Money(300), Money(800)]print("Cheapest:", min(prices))print("Most expensive:", max(prices))
```

## Practice Questions### Q1: Time ClassCreate a `Time` class with `hours` and `minutes`.Implement:- `__str__` → `"HH:MM"` format (zero-padded)- `__add__` → add two Time objects (handle carry-over at 60 min / 24 hr)- `__sub__` → subtract time (result always positive — absolute difference)- `__lt__`, `__eq__` → for sorting/comparison- `__gt__`Test: `10:45 + 2:30`, `08:00 - 09:30` (abs diff), sort a list of times

```python
# Q1 Solution: Time Classclass Time:    def __init__(self, hours, minutes):        total       = hours * 60 + minutes        self.hours  = (total // 60) % 24        self.minutes = total % 60    def to_minutes(self):        return self.hours * 60 + self.minutes    def __str__(self):        return f"{self.hours:02d}:{self.minutes:02d}"    def __repr__(self):        return f"Time({self.hours}, {self.minutes})"    def __add__(self, other):        total = self.to_minutes() + other.to_minutes()        return Time(0, total)    def __sub__(self, other):        diff = abs(self.to_minutes() - other.to_minutes())        return Time(0, diff)    def __eq__(self, other):        return self.to_minutes() == other.to_minutes()    def __lt__(self, other):        return self.to_minutes() < other.to_minutes()    def __gt__(self, other):        return self.to_minutes() > other.to_minutes()t1 = Time(10, 45)t2 = Time(2,  30)t3 = Time(8,   0)t4 = Time(9,  30)print(f"{t1} + {t2} = {t1 + t2}")print(f"|{t3} - {t4}| = {t3 - t4}")print(f"{t1} > {t3}: {t1 > t3}")times = [Time(14, 30), Time(9, 0), Time(22, 15), Time(6, 45)]times.sort()print("Sorted:", [str(t) for t in times])
```

### Q2: Custom List (Bag)Create class `Bag` (like a multiset):- `__init__` — empty dictionary `{item: count}`- `add(item)` — increment count- `remove(item)` — decrement, raise error if not present- `__add__` — merge two bags (sum counts)- `__contains__` — check if item in bag- `__len__` — total number of items (sum of all counts)- `__str__` — display contents

```python
# Q2 Solution: Bag (Multiset)class Bag:    def __init__(self):        self._data = {}    def add(self, item):        self._data[item] = self._data.get(item, 0) + 1    def remove(self, item):        if item not in self._data:            raise KeyError(f"'{item}' not in bag")        self._data[item] -= 1        if self._data[item] == 0:            del self._data[item]    def __add__(self, other):        result = Bag()        for item, count in self._data.items():            result._data[item] = count        for item, count in other._data.items():            result._data[item] = result._data.get(item, 0) + count        return result    def __contains__(self, item):        return item in self._data    def __len__(self):        return sum(self._data.values())    def __str__(self):        items = ", ".join(f"{k}x{v}" for k, v in sorted(self._data.items()))        return f"Bag({{{items}}}) [total: {len(self)}]"b1 = Bag()b1.add("apple"); b1.add("banana"); b1.add("apple")b2 = Bag()b2.add("apple"); b2.add("cherry")print(b1)print(b2)print("apple in b1:", "apple" in b1)print("Combined:", b1 + b2)b1.remove("apple")print("After removing one apple:", b1)
```

## Key Takeaways- Operator overloading makes custom classes feel native and intuitive- Always return a **new object** from `__add__`, `__mul__` etc. (don't mutate)- Implement comparison methods (`__lt__`, `__eq__`) to enable `sort()` and `min()`/`max()`- Currency/unit-safety checks in arithmetic methods prevent logic bugs
